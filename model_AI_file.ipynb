```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn import datasets
```

```python
df=datasets.load_breast_cancer(as_frame=True)
```

```python
data=df.data
target=df.target
data['target']=target
pd.set_option('display.max_columns',None)
```

```python
data.head()
```

```python
df['target'].value_counts()
```

```python
X= data.drop('target',axis=1)   ##always be in dataFrame
y=data['target']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=417)
```

```python
from sklearn.svm import LinearSVC
```

```python
model = LinearSVC(penalty="l2",loss='squared_hinge',C=10)
```

```python
model.fit(X_train,y_train)
```
